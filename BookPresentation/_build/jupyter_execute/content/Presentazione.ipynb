<div style="font-size: 100px">
Fake news detected
</div>

<div style="font-size: 50px">
Obiettivo
</div>

### L'obiettivo di questo progetto è sviluppare un sistema avanzato in grado di analizzare in tempo reale un flusso di notizie provenienti da diverse fonti (feed RSS, API di notizie, social media) e determinare se una notizia è potenzialmente falsa utilizzando l'API.

<div style="font-size: 50px">
Perché serve un sistema di Fake News Detection?
</div>

## - **Proliferazione di contenuti non verificati**: Con l'avvento dei social media e dei blog, chiunque può pubblicare notizie senza un controllo editoriale.

## - **Conseguenze della disinformazione**: Impatto sull'opinione pubblica, rischi per la democrazia, tensioni sociali e danni alla reputazione di persone o aziende.

## - **Difficoltà nel riconoscere le informazioni attendibili**: Gli utenti faticano a distinguere una notizia vera da una manipolata o completamente falsa.

<div style="font-size: 50px">
Pro
</div>

## - **Supporto decisionale**: Potrebbe essere usato da giornalisti, ricercatori e utenti generali per valutare la veridicità delle informazioni in modo rapido ed efficiente.

## - **Fornire affidabilità immediata**: Questo sistema potrebbe aiutare gli utenti a identificare rapidamente notizie false, riducendo la diffusione di disinformazione.

## - **Aggiornamento continuo**: Un sistema che monitora i feed RSS, le API di notizie e i social media sarebbe in grado di adattarsi velocemente a nuovi sviluppi e nuove fonti di disinformazione

<div style="font-size: 50px">
Contro
</div>

## - **Richiesta di risorse elevate**: L'analisi in tempo reale richiede una notevole capacità di calcolo e infrastrutture adeguate, soprattutto per gestire grandi flussi di notizie e dati in tempo reale. Questo potrebbe comportare costi elevati per il mantenimento del sistema.

## - **Dipendenza dalle fonti terze**: L'API di Google FactCheck è uno strumento potente, ma dipendere completamente da un singolo servizio di fact-checking potrebbe non coprire tutti gli aspetti o le voci di disinformazione. Alcune notizie potrebbero non essere controllate o verificate

## - **Affidabilità dei dati di input**: Il sistema si basa su fonti esterne (RSS, API di notizie, social media), che possono essere parziali o non sempre accurate. Se i dati in ingresso sono già distorti o incompleti, l'analisi potrebbe risultare errata.
